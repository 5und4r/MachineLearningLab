## Apriori Algorithm Implementiation


## The Apriori Algorithm

The Apriori algorithm is a foundational data mining technique used for frequent itemset mining and association rule learning over transactional databases. Introduced by Rakesh Agrawal and Ramakrishnan Srikant in 1994, it is most famous for powering Market Basket Analysis to discover hidden relationships between items (e.g., "If a customer buys bread, they are 80% likely to also buy butter").

### Core Principle of Apriori

The algorithm significantly cuts down computation time using the Apriori Property (or down-closure property of support):

Frequent Rule:

If an itemset is frequent, all of its subsets must also be frequent.

Infrequent (Pruning) Rule:

If an itemset is infrequent, all of its supersets will automatically be infrequent. For example, if {Beer} is rarely bought, there is no need to waste time checking {Beer, Diapers} or {Beer, Diapers, Chips}


### How Apriori Works:

The algorithm uses a "bottom-up" approach, progressing iteratively from single items to larger groups:

Step 1: Generate 1-Itemsets (C₁) – Scan the database to count the frequency of every single individual item.

Step 2: Prune – Eliminate all items that fall below a user-defined Minimum Support threshold. The survivors become the frequent 1-itemsets (L₁).

Step 3: Join Step – Combine L₁ items with each other to generate pairs, known as candidate 2-itemsets (C₂).

Step 4: Prune Again – Scan the database to calculate the support of these pairs and filter out the infrequent ones to get L₂.

Step 5: Repeat – Continue joining and pruning to generate 3-itemsets (C₃, L₃), 4-itemsets, etc., until no more frequent itemsets can be formed.

Step 6: Rule Generation – Take the final frequent itemsets and extract rules that meet a user-defined Minimum Confidence threshold.

## Implementation using python

````python
from itertools import combinations

def calculate_support(transactions, itemset):
    count = 0
    for t in transactions:
        if itemset.issubset(t):
            count += 1
    return count

def run_apriori(transactions, min_support, question_label):
    print(f"\n=============================================")
    print(f"   APRIORI ALGORITHM FOR {question_label.upper()}   ")
    print(f"=============================================\n")

    # Extract unique items
    unique_items = set(item for t in transactions for item in t)

    # --- STEP 1: Candidate 1-itemsets (C1) & Frequent 1-itemsets (L1) ---
    C1 = {frozenset([item]): calculate_support(transactions, {item}) for item in unique_items}
    print(" Candidate 1-itemsets (C1):")
    for item, sup in sorted(C1.items()):
        print(f"  {set(item)} -> Support Count: {sup}")

    L1 = {item: sup for item, sup in C1.items() if sup >= min_support}
    print("\n Frequent 1-itemsets (L1):")
    for item, sup in sorted(L1.items()):
        print(f"  {set(item)} -> Support Count: {sup}")

    if not L1:
        print("No frequent itemsets found. Stopping.")
        return

    # --- STEP 2: Candidate 2-itemsets (C2) & Frequent 2-itemsets (L2) ---
    l1_lists = list(L1.keys())
    C2 = {}
    for i in range(len(l1_lists)):
        for j in range(i + 1, len(l1_lists)):
            candidate = l1_lists[i].union(l1_lists[j])
            C2[candidate] = calculate_support(transactions, candidate)

    print("\n Candidate 2-itemsets (C2):")
    for item, sup in sorted(C2.items()):
        print(f"  {set(item)} -> Support Count: {sup}")

    L2 = {item: sup for item, sup in C2.items() if sup >= min_support}
    print("\n Frequent 2-itemsets (L2):")
    for item, sup in sorted(L2.items()):
        print(f"  {set(item)} -> Support Count: {sup}")

    # --- STEP 3: Candidate 3-itemsets (C3) with Join & Prune ---
    l2_lists = list(L2.keys())
    joined_candidates = set()

    # Join Step
    for i in range(len(l2_lists)):
        for j in range(i + 1, len(l2_lists)):
            cand = l2_lists[i].union(l2_lists[j])
            if len(cand) == 3:
                joined_candidates.add(cand)

    # Prune Step
    pruned_candidates = []
    print("\n Apriori Pruning Process for C3:")
    for cand in joined_candidates:
        subsets = [frozenset(c) for c in combinations(cand, 2)]
        is_valid = True
        for sub in subsets:
            if sub not in L2:
                print(f"   Pruning {set(cand)} because its subset {set(sub)} is not frequent (not in L2).")
                is_valid = False
                break
        if is_valid:
            pruned_candidates.append(cand)

    C3 = {cand: calculate_support(transactions, cand) for cand in pruned_candidates}

    print("\n Candidate 3-itemsets (C3) After Pruning:")
    if not C3:
        print("  [Empty Set]")
    for item, sup in sorted(C3.items()):
        print(f"  {set(item)} -> Support Count: {sup}")

    # Frequent 3-itemsets (L3)
    L3 = {item: sup for item, sup in C3.items() if sup >= min_support}
    print("\n Frequent 3-itemsets (L3):")
    if not L3:
        print("  [Empty Set - No 3-itemsets satisfy min_support]")
    for item, sup in sorted(L3.items()):
        print(f"  {set(item)} -> Support Count: {sup}")




````

````python
# Data for Question 1: Hospital Pharmacy Analysis
pharmacy_transactions = [
    {"Paracetamol", "Vitamin C", "Cough Syrup"},
    {"Paracetamol", "Antibiotic", "Vitamin C"},
    {"Antibiotic","Vitamin C", "Pain Reliever"},
    {"Paracetamol", "Vitamin C", "Pain Reliever"},
    {"Paracetamol", "Antibiotic", "Vitamin C"},
    {"Vitamin C", "Cough Syrup"}
]

run_apriori(pharmacy_transactions, min_support=3, question_label="Question 1: Hospital Pharmacy")
````

## How the code works


###1. The Support Counter (calculate_support)

This is the workhorse helper function. It takes any group of items (like a combination of books or medicines) and counts how many times that exact combination appears together across the entire list of transactions using the .issubset() check.

###2. Phase 1: Finding Frequent Individual Items ($L_1$)

The code extracts every unique individual item from the data.It counts their individual frequencies.It immediately drops any item that falls below the min_support threshold of 3, ensuring the code doesn't waste time on rare items.

###3. Phase 2: Building and Filtering Pairs ($L_2$)

Using the items that survived the first phase, it uses nested loops to pair them up (the Join step).It checks the frequency of these pairs against the database and filters out those with a count lower than 3.

###4. Phase 3: The Optimized 3-Itemset Check ($L_3$)

This phase showcases the true power of the Apriori algorithm by utilizing two distinct steps before hitting the database:

The Join Step: It combines the surviving pairs from $L_2$ to form potential 3-item combinations.

The Prune Step: For each new 3-item candidate, it generates all possible 2-item subsets. It checks if all of those subsets exist in $L_2$. If even one subset is missing, the code discards the candidate immediately. This prevents the program from performing unnecessary database scans on combinations destined to fail.Final Scan: It counts the support only for the candidates that survived pruning to output the final frequent 3-itemsets.

# FREQUENT PATTERN TREE IMPLEMENTATION

## Frequent Pattern tree (FP) Growth

The FP-Growth (Frequent Pattern Growth) algorithm is an efficient data mining method used to find frequent itemsets in a transactional database without generating candidate sets. Developed by Jiawei Han in 2000, it solves the core bottleneck of the older ⁠Apriori algorithm by compressing the database into a compact tree structure called an FP-Tree and scanning the database only twice.

## How FP Growth works:

The algorithm follows a two phase approach:

### Phase 1: Building the FP tree

1. First Scan:Scan the database to count the support (frequency) of every individual item. Discard items that fall below the user-defined Minimum Support Threshold.

2. Sort and Prioritize: Order the remaining frequent items in descending order of their support count to create a Header Table.

3. Second Scan:
Scan the database again. Rearrange the items within each transaction so they match the descending order of the Header Table.

4. Insert into Tree:
Insert these ordered transactions into a prefix tree starting from a null root. If multiple transactions share the same starting items, they share the same tree nodes, and the node counters increment by 1. Diverging items split into new branches.

### Phase 2: Mining frequent patterns:

Instead of analyzing the raw database, the algorithm recursively extracts patterns directly from the built FP-Tree starting from the least frequent item at the bottom:
1. Conditional Pattern Base: Find all prefix paths in the tree leading to a specific item.

2. Conditional FP-Tree: Build a small, temporary FP-tree using only the item counts found in those paths.

3. Pattern Generation: Generate frequent pairs and rules by combining the item with the frequent elements in its conditional tree.



## Implementation using Python

````python
from collections import Counter

def build_compact_fp_tree(transactions, min_support, scenario_name):
    # 1. Count frequencies and keep only items meeting min_support
    counts = Counter(item for trans in transactions for item in trans)
    freq_items = {item: c for item, c in counts.items() if c >= min_support}

    # Create a priority rank: sort by frequency (highest first), then alphabetically
    rank = {item: i for i, (item, _) in enumerate(sorted(freq_items.items(), key=lambda x: (-x[1], x[0])))}

    # 2. Build the Tree using a nested dictionary structure: { item: [count, {children}] }
    root = {}
    for trans in transactions:
        current_node = root
        # Filter out infrequent items and sort the transaction by priority rank
        sorted_trans = sorted([item for item in trans if item in freq_items], key=lambda x: rank[x])

        for item in sorted_trans:
            if item not in current_node:
                current_node[item] = [0, {}]  # Format: [Count, Children Dictionary]
            current_node[item][0] += 1        # Step 1: Increment count
            current_node = current_node[item][1] # Step 2: Move deeper into the children dict

    # 3. Ultra-simple recursive print function
    def display_tree(node, indent=1):
        for item, (count, children) in node.items():
            print("  " * indent + f"└── {item}: {count}")
            display_tree(children, indent + 1)

    print(f"\n FP-TREE FOR: {scenario_name}")
    display_tree(root)



````

````python
#minimum support = 3
movie_trans = [
    ["Action", "Comedy", "Thriller"], ["Action", "Drama", "Thriller"],
    ["Comedy", "Thriller", "Romance"], ["Action", "Comedy", "Thriller"],
    ["Action", "Comedy", "Drama"]
]
build_compact_fp_tree(movie_trans, 3, "Scenario 1: Movie Streaming")
````

````python
iot_trans = [
    ["Smart Light", "Smart TV", "Air Conditioner"], ["Smart Light", "Smart Speaker", "Air Conditioner"],
    ["Smart TV", "Air Conditioner", "Smart Speaker"], ["Smart Light", "Smart TV", "Air Conditioner"],
    ["Smart Light", "Air Conditioner", "Smart Speaker"]
]
build_compact_fp_tree(iot_trans, 3, "Scenario 2: Smart Home IoT")
````

## Working of the code

This program implements the Frequent Pattern (FP) Tree Growth algorithm from scratch to find frequent itemsets for both the Movie Streaming and Smart Home IoT scenarios using a minimum support count threshold of 3. Instead of using complex object-oriented classes and pointer configurations, this code uses a highly efficient, nested Python dictionary architecture.

### Data structure architecture:

Each tree node is represented as a simple two-element list inside a dictionary:$$\text{Node} = [\text{Count}, \{\text{Children Dictionary}\}]$$This allows branches to nest deeply within each other in memory. For example, a shared path looks like this behind the scenes:

    ```{ 'Action': [4, { 'Comedy': [3, { 'Thriller': [2, {}] }] }] }```


### Execution steps:

1. Step 1: Frequency Counting & Ranking The code uses Counter to calculate the total occurrence of each individual item across all transactions. Items failing to meet the minimum support count of 3 are immediately dropped. The surviving items are sorted in descending order of frequency to create a priority ranking list.

2. Step 2: Transaction Sorting & Filtering The code loops through each transaction row, filters out the infrequent items, and sorts the remaining items based on their priority rank. Sorting ensures that the most frequent items are placed near the root node, maximizing path-sharing between different transactions.

3. Step 3: Building Tree Paths The code traverses each sorted transaction line item by line item starting from a blank root = {} dictionary:If an item does not exist at the current branch level, a new node initialized to [0, {}] is created.The path's frequency count (index 0) is incremented by 1.The position pointer shifts directly inside that item's sub-dictionary of children (index 1), ready to process the next item in the sequence.

4. Step 4: Recursive Tree Visualization To print the final tree layout clearly, a recursive function (display_tree) unpacks each node's count and sub-dictionary. It prints the item name alongside its execution count, multiplying string indentations ("  " * indent) to visually render the branch depths beautifully in the terminal window.

# MULTIDIMENSIONAL FREQUENT PATTERN MINING AND CUBOID VISUALIZATION

## Dataset chosen:
 F1 Strategy dataset for pit stop prediction.

## 1. Preprocessing and One Hot Encoding

````python
import pandas as pd

# Step 1: Define our 5 dimensions
dimensions = ['Driver', 'Compound', 'Race', 'Stint', 'PitStop']

# Extract dimensions into a new DataFrame
cube_df = df[dimensions].copy()

# Drop the few missing values in the 'Compound' column
cube_df.dropna(inplace=True)

# Format the numeric columns so they look like categories in Association rules

cube_df['Stint'] = 'Stint_' + cube_df['Stint'].astype(str)
cube_df['PitStop'] = cube_df['PitStop'].apply(lambda x: 'Pit_Yes' if x == 1 else 'Pit_No')

# we take a random sample of 2,000 records
cube_sample = cube_df.sample(n=2000, random_state=42)

# Step 2: Convert to One-Hot Encoded format
one_hot_df = pd.get_dummies(cube_sample, dtype=bool)

print(f"Selected dimensions: {dimensions}")
print(f"Shape of encoded dataset: {one_hot_df.shape}")
print("\nFirst 5 rows of One-Hot Encoded Dataset:")
display(one_hot_df.head())
````

````python
from mlxtend.frequent_patterns import apriori, association_rules

# Step 1: Apply the Apriori algorithm
# min_support=0.05 means a pattern must appear in at least 5% of the 2,000 rows
frequent_itemsets = apriori(one_hot_df, min_support=0.05, use_colnames=True)
print(f"Discovered {len(frequent_itemsets)} frequent itemsets.")

# Step 2: Generate Association Rules
#Generating Association with 50% confidence
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

# Step 3: Sort the rules by Lift (to find the most interesting/meaningful ones)
rules = rules.sort_values('lift', ascending=False)

print(f"Generated {len(rules)} association rules.\n")

print("Top 10 Association Rules:")
# Displaying just the most relevant columns for readability
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
````
